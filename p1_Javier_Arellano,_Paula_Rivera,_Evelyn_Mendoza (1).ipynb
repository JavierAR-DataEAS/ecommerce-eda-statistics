```python
# Configurar los colores usando la misma paleta 'mako'
colors = sns.color_palette('mako', len(fi))

# Reducir ligeramente el tamaño de la figura para mejor visibilidad
plt.figure(figsize=(9, 7))

# Crear la gráfica de pastel con etiquetas más claras
plt.pie(fi.values, labels=fi.index, autopct='%1.1f%%', startangle=140,
        colors=colors, pctdistance=0.80, textprops={'fontsize': 9})

# Dibujar un círculo blanco en el centro (estilo dona)
centre_circle = plt.Circle((0,0), 0.65, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

# Título manteniendo su formato
plt.title('Distribución Porcentual por País', fontsize=14, fontweight='bold', color='#333333', pad=10)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
# Seleccionar 10 datos aleatorios
sample_age = df['age'].sample(10, random_state=42)

# Calcular media y desviación estándar
mean_age = df['age'].mean()
std_age = df['age'].std()

# Calcular valor Z: z = (x - mean) / std
z_scores = (sample_age - mean_age) / std_age

# Crear un DataFrame para mostrar los resultados comparativos
z_results = pd.DataFrame({
    'Edad Original': sample_age,
    'Valor Z': z_scores
})

display(z_results)
```
